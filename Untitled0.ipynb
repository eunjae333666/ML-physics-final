```python
class LJPotential(nnx.Module):
    def __init__(self, sigma=1.0, epsilon=1.0, wall_k=20.0):
        self.sigma = sigma
        self.epsilon = epsilon
        self.wall_k = wall_k

    def __call__(self, x, Z=None, V=None, progress=1.0):
        if x.ndim == 1:
            x = x.reshape(-1, 3)

        N = x.shape[0]

        if hasattr(V, 'ndim') and V.ndim > 0:
            V_val = jnp.squeeze(V)
        else:
            V_val = V

        diff = x[:, None, :] - x[None, :, :]
        r2 = jnp.sum(diff**2, axis=-1)

        mask = 1.0 - jnp.eye(N)
        r2_safe_diag = r2 + (1.0 - mask) # 대각 성분 0 방지용

        # ---------------------------------------------------------
        # 🚨 [수정된 파트: 정밀한 Soft-core Annealing 적용]
        # ---------------------------------------------------------
        # progress가 0일 때 alpha=0.5 (매끄러움), 1일 때 alpha=0.001 (실제 LJ에 근접)
        alpha = jnp.interp(progress, jnp.array([0.0, 0.8]), jnp.array([0.5, 0.001]))

        # r^6 계산 (대각선은 위에서 1.0을 더해줬으므로 안전함)
        r6 = (r2_safe_diag / self.sigma**2) ** 3

        # Soft-core 분모 적용 (0으로 나누어지는 것 원천 차단)
        denom = alpha + r6
        inv_r6_soft = 1.0 / denom

        # 새로운 LJ 매트릭스 계산
        lj_matrix = 4.0 * self.epsilon * (inv_r6_soft**2 - inv_r6_soft)

        # 에너지 폭발을 막기 위한 클리핑.
        # 분모(alpha) 덕분에 이미 부드러워졌으므로 극단적인 클리핑(100,000) 불필요
        U_0 = 1000.0 # 스케일링 기준점

        # 1. 일단 마스크를 씌워서 순수 LJ 에너지를 모두 합칩니다.
        U_raw = 0.5 * jnp.sum(mask * lj_matrix)
        # ---------------------------------------------------------
        safe_U_raw = jnp.maximum(U_raw, U_0)
        U_eff = jnp.where(
            U_raw < U_0,
            U_raw, # 정상 에너지(Bulk)는 물리학 그대로 통과
            U_0 * (1.0 + jnp.log(safe_U_raw / U_0)) # 튄 에너지(Tail)는 로그로 부드럽게 압축
        )

        # 정육면체 큐브 벽
        L_box = V_val ** (1/3)
        half_L = L_box / 2.0
        out_of_bounds = jnp.maximum(jnp.abs(x) - half_L, 0.0)
        U_wall = self.wall_k * jnp.sum(out_of_bounds ** 4)

        return U_eff + U_wall
```

### 2.1. flow 순수 계산 부분

```python
# @title
class ConditionalAffineCoupling(nnx.Module):
    # mask[i] == 1  ->  dimension i is "kept"
    # mask[i] == 0  ->  dimension i is "transformed"
    #절반만 바꿔야하니까... 아무래도
    def __init__(self, d, hidden, mask, cond_dim=2, *, rngs):
        self.mask = mask
        # The s and t networks are an ordinary MLP. They look at the kept dims AND conditions
        # and produce a scaling and a shift for the transformed dims.
        self.net = nnx.Sequential(
            nnx.Linear(d + cond_dim, hidden, rngs=rngs), nnx.silu, #T,V에 따라 변환이 다르게 되어야 하므로 +cond_dim을 해줌.
            nnx.Linear(hidden, hidden, rngs=rngs), nnx.silu, # relu는 부드럽지 않아서 2차 미분 계산(metric)시 망할 수 있음.
            nnx.Linear(hidden, 2 * d, kernel_init=nnx.initializers.zeros_init(), rngs=rngs),   # 결국 내보내는건 T,V를 제외한 x 각각의 s, t임.(!2*(d+cond_dim))
        )
        self.d = d

    def _scale_and_shift(self, kept, t_cond, v_cond):
        # 차원 맞추기: T,V가 걍 숫자로 들어오면, x(kept)의 형식이랑 맞춰주기.
        t_c = jnp.atleast_1d(t_cond) #숫자를 1차원 배열이라도 만들기
        v_c = jnp.atleast_1d(v_cond)
        if t_c.ndim == 1: t_c = t_c[:, None] #kept의 형식에 맞추기(열 벡터? 형태인듯)
        if v_c.ndim == 1: v_c = v_c[:, None]

        # kept 차원에 t_cond, v_cond를 이어 붙임 (d + 2 차원이 됨)
        # 그냥 숫자로 들어왔을 경우 그 값 복사해서 크기 맞춰줌
        cond_input = jnp.concatenate([kept, t_c, v_c], axis=-1)

        st = self.net(cond_input) #net.__call__(cond_input)을 일케 줄여서 쓸 수 있는거임. 그니까 Sequential이라는 class가 있다고.
        s_raw, t = st[..., :self.d], st[..., self.d:]
        s = jnp.tanh(s_raw) * 2.0   # bound |s| < 2 for training stability
        return s, t

    def forward(self, z, t_cond, v_cond):
        # Input z, output x = (z_A, z_B * exp(s(z_A)) + t(z_A)).
        s, t = self._scale_and_shift(z * self.mask, t_cond, v_cond) #z_A에 대해 s,t뽑아냄.(s,t의 차원은 z_B)
        transformed = 1.0 - self.mask
        x = z * jnp.exp(s * transformed) + t * transformed #transformed==0(A부분)이면, x=z_A.
        log_det = jnp.sum(s * transformed, axis=-1) # log(det J)=sum s
        return x, log_det

    def inverse(self, x, t_cond, v_cond):
        # Input x, output z = (x_A, (x_B - t) * exp(-s)).
        s, t = self._scale_and_shift(x * self.mask, t_cond, v_cond)
        transformed = 1.0 - self.mask
        z = (x - t * transformed) * jnp.exp(-s * transformed)
        log_det = -jnp.sum(s * transformed, axis=-1)
        return z, log_det


class ConditionalRealNVP(nnx.Module):
    # A stack of coupling layers with alternating masks.
    def __init__(self, d=96, n_layers=16, hidden=128, cond_dim=2, *, rngs):
        masks = []
        for i in range(n_layers): #각 layer에서 A,B 부분 번갈아가면서 변환 시켜야 하니...
            mask = jnp.zeros(d)
            mask = mask.at[:d//2].set(1.0) if i % 2 == 0 else mask.at[d//2:].set(1.0)
            masks.append(mask)
        # 하위 레이어 생성 시 cond_dim을 전달
        self.layers = [ConditionalAffineCoupling(d, hidden, m, cond_dim, rngs=rngs) for m in masks]
        self.d = d

    def forward(self, z, t_cond, v_cond):
        # Push z through every layer, accumulating log|det J|.
        log_det = jnp.zeros(z.shape[0]); x = z
        for layer in self.layers:
            x, ld = layer.forward(x, t_cond, v_cond); log_det += ld
        return x, log_det

    def inverse(self, x, t_cond, v_cond):
        log_det = jnp.zeros(x.shape[0]); z = x
        for layer in reversed(self.layers):
            z, ld = layer.inverse(z, t_cond, v_cond); log_det += ld
        return z, log_det

    def log_prob(self, x, t_cond, v_cond):
        """ log p_theta(x) 구하는 함수"""
        # log p(x) = log p_Z(z) + log|det J_{f^{-1}}(x)|, with z = f^{-1}(x).
        z, log_det_inv = self.inverse(x, t_cond, v_cond)
        log_pz = -0.5 * jnp.sum(z**2, axis=-1) - 0.5 * self.d * jnp.log(2 * jnp.pi)
        return log_pz + log_det_inv

    def sample(self, key, n, t_cond, v_cond):
        z = jr.normal(key, (n, self.d))
        x, _ = self.forward(z, t_cond, v_cond)
        return x
```

```python
def rational_quadratic_spline_1d(x, w, h, d, inverse=False, B=1.0):
    """
    정규화된 단일 차원에 대한 Rational Quadratic Spline (RQS) 변환
    - 주의사항 1 반영: 하한선 필터 및 경계 밖 (|x| > B) 고속도로(Identity Tail) 적용
    """
    # 1. 상자 바깥 영역 방어막 (|x| > B 이면 변환 없이 기울기 1로 관통)
    inside_mask = (x > -B) & (x < B)

    # 안전한 연산을 위해 경계 밖의 값들을 임시로 상자 안으로 클리핑 (나중에 마스킹 처리)
    x_clamped = jnp.clip(x, -B + 1e-5, B - 1e-5)

    # 매듭점(Knot) 너비와 높이 설정 (Softmax를 통해 합이 2B가 되도록 분배)
    K = w.shape[-1]
    widths = jax.nn.softmax(w, axis=-1) * (2.0 * B)
    heights = jax.nn.softmax(h, axis=-1) * (2.0 * B)

    # 매듭점의 절대 좌표 생성 (-B부터 시작)
    knot_x = -B + jnp.cumsum(jnp.pad(widths, ((0,0), (1,0))), axis=-1)
    knot_y = -B + jnp.cumsum(jnp.pad(heights, ((0,0), (1,0))), axis=-1)

    # 기울기(Derivatives) 제어: 미분값 최솟값 하한선 1e-3 강제 (주의사항 1)
    # 경계점(맨 첫점과 끝점)의 기울기는 1.0으로 고정하여 선형 꼬리(Identity Tail)와 매끄럽게 연결
    internal_derivatives = jax.nn.softplus(d) + 1e-3
    derivatives = jnp.pad(internal_derivatives, ((0,0), (1,1)), constant_values=1.0)

    # 현재 x가 속한 빈(Bin)의 인덱스 찾기
    if not inverse:
        bin_idx = jnp.clip(jnp.searchsorted(knot_x, x_clamped, side='right') - 1, 0, K - 1)
    else:
        bin_idx = jnp.clip(jnp.searchsorted(knot_y, x_clamped, side='right') - 1, 0, K - 1)

    # 인덱스별 매듭점 정보 추출
    gather_w = lambda arr, idx: jnp.take_along_axis(arr, idx[..., None], axis=-1)[..., 0]

    x_k = gather_w(knot_x, bin_idx)
    x_k1 = gather_w(knot_x, bin_idx + 1)
    y_k = gather_w(knot_y, bin_idx)
    y_k1 = gather_w(knot_y, bin_idx + 1)

    d_k = gather_w(derivatives, bin_idx)
    d_k1 = gather_w(derivatives, bin_idx + 1)

    w_b = x_k1 - x_k
    h_b = y_k1 - y_k
    s_b = h_b / w_b

    # RQS 수식 계산
    if not inverse:
        xi = (x_clamped - x_k) / w_b
        xi_1_xi = xi * (1.0 - xi)

        num = h_b * (s_b * xi**2 + d_k * xi_1_xi)
        den = s_b + (d_k1 + d_k - 2.0 * s_b) * xi_1_xi
        y_hat = y_k + num / den

        # 야코비안 계산
        num_deriv = s_b**2 * (d_k1 * xi**2 + 2.0 * s_b * xi_1_xi + d_k * (1.0 - xi)**2)
        den_deriv = (s_b + (d_k1 + d_k - 2.0 * s_b) * xi_1_xi)**2
        dk_dx = num_deriv / den_deriv
        log_det_hat = jnp.log(dk_dx)

        # 마스킹 조합 (안쪽은 스플라인, 바깥쪽은 Identity 관통)
        final_out = jnp.where(inside_mask, y_hat, x)
        final_logdet = jnp.where(inside_mask, log_det_hat, 0.0)
        return final_out, final_logdet
    else:
        # 역변환 계산
        y_clamped = x_clamped
        y_delta = y_clamped - y_k

        a = h_b * (s_b - d_k) + y_delta * (d_k1 + d_k - 2.0 * s_b)
        b = h_b * d_k - y_delta * (d_k1 + d_k - 2.0 * s_b)
        c = -s_b * y_delta

        # 2차 방정식 해 구하기
        rhs = b**2 - 4.0 * a * c
        xi = (2.0 * c) / (-b - jnp.sqrt(jnp.maximum(rhs, 1e-6)))

        x_hat = xi * w_b + x_k

        # 역변환 야코비안 계산
        xi_1_xi = xi * (1.0 - xi)
        num_deriv = s_b**2 * (d_k1 * xi**2 + 2.0 * s_b * xi_1_xi + d_k * (1.0 - xi)**2)
        den_deriv = (s_b + (d_k1 + d_k - 2.0 * s_b) * xi_1_xi)**2
        dk_dx = num_deriv / den_deriv
        log_det_hat = -jnp.log(dk_dx)

        final_out = jnp.where(inside_mask, x_hat, x)
        final_logdet = jnp.where(inside_mask, log_det_hat, 0.0)
        return final_out, final_logdet


class ConditionalSplineCoupling(nnx.Module):
    def __init__(self, d, hidden, mask, n_bins=8, cond_dim=2, *, rngs):
        self.mask = mask  # 주의사항 3에 의해 [N, 3] 구조를 보존하는 마스크
        self.d = d
        self.n_bins = n_bins

        # 변환될 차원 수 계산
        n_transformed = int(d - jnp.sum(mask))

        # 각 변환 차원당 필요한 매듭점 파라미터 수: 너비(K) + 높이(K) + 내부 유도치(K-1) = 3K - 1
        self.out_per_dim = 3 * n_bins - 1
        self.n_transformed = n_transformed

        # 파라미터를 뱉어낼 MLP 설계
        self.net = nnx.Sequential(
            nnx.Linear(d + cond_dim, hidden, rngs=rngs), nnx.silu,
            nnx.Linear(hidden, hidden, rngs=rngs), nnx.silu,
            nnx.Linear(hidden, n_transformed * self.out_per_dim, kernel_init=nnx.initializers.zeros_init(), rngs=rngs)
        )

    def _predict_spline_params(self, kept, t_cond, v_cond):
        t_c = jnp.atleast_1d(t_cond)[..., None] if jnp.atleast_1d(t_cond).ndim == 1 else jnp.atleast_1d(t_cond)
        v_c = jnp.atleast_1d(v_cond)[..., None] if jnp.atleast_1d(v_cond).ndim == 1 else jnp.atleast_1d(v_cond)
        cond_input = jnp.concatenate([kept, t_c, v_c], axis=-1)

        outputs = self.net(cond_input)
        # (batch, n_transformed, 3K - 1) 형태로 분할
        outputs = outputs.reshape(outputs.shape[0], self.n_transformed, self.out_per_dim)

        w = outputs[..., :self.n_bins]
        h = outputs[..., self.n_bins:2*self.n_bins]
        d = outputs[..., 2*self.n_bins:]
        return w, h, d

    def forward(self, z, t_cond, v_cond):
        # 고정할 그룹 데이터 추출
        kept_data = z * self.mask
        w, h, d = self._predict_spline_params(kept_data, t_cond, v_cond)

        # 변환 그룹 데이터 순치 배열 추출
        transformed_indices = jnp.where(self.mask == 0.0)[0]
        z_trans = z[:, transformed_indices]

        # 각 차원에 대해 완전히 부드러운 스플라인 전방향 사상 진행
        x_trans_list = []
        logdet_list = []
        for i in range(self.n_transformed):
            x_i, ld_i = rational_quadratic_spline_1d(z_trans[:, i], w[:, i], h[:, i], d[:, i], inverse=False, B=1.0)
            x_trans_list.append(x_i)
            logdet_list.append(ld_i)

        x_trans = jnp.stack(x_trans_list, axis=-1)
        log_det = jnp.sum(jnp.stack(logdet_list, axis=-1), axis=-1)

        # 원본 구조로 복원
        x = z.at[:, transformed_indices].set(x_trans)
        return x, log_det

    def inverse(self, x, t_cond, v_cond):
        kept_data = x * self.mask
        w, h, d = self._predict_spline_params(kept_data, t_cond, v_cond)

        transformed_indices = jnp.where(self.mask == 0.0)[0]
        x_trans = x[:, transformed_indices]

        z_trans_list = []
        logdet_list = []
        for i in range(self.n_transformed):
            z_i, ld_i = rational_quadratic_spline_1d(x_trans[:, i], w[:, i], h[:, i], d[:, i], inverse=True, B=1.0)
            z_trans_list.append(z_i)
            logdet_list.append(ld_i)

        z_trans = jnp.stack(z_trans_list, axis=-1)
        log_det = jnp.sum(jnp.stack(logdet_list, axis=-1), axis=-1)

        z = x.at[:, transformed_indices].set(z_trans)
        return z, log_det


class ConditionalNSF(nnx.Module):
    def __init__(self, d=96, n_layers=8, hidden=128, n_bins=8, cond_dim=2, *, rngs):
        masks = []
        for i in range(n_layers):
            mask = jnp.zeros(d)
            if i % 2 == 0:
                mask = mask.at[:d//2].set(1.0)  # 전반부 N/2개 입자 보존, 후반부 변환
            else:
                mask = mask.at[d//2:].set(1.0)  # 후반부 N/2개 입자 보존, 전반부 변환
            masks.append(mask)

        self.layers = [ConditionalSplineCoupling(d, hidden, m, n_bins, cond_dim, rngs=rngs) for m in masks]
        self.d = d

    def forward(self, z, t_cond, v_cond):
        # 표준 상자 [-1, 1] 안에서만
        log_det = jnp.zeros(z.shape[0])
        x = z
        for layer in self.layers:
            x, ld = layer.forward(x, t_cond, v_cond)
            log_det += ld
        return x, log_det

    def inverse(self, x, t_cond, v_cond):
        log_det = jnp.zeros(x.shape[0])
        z = x
        for layer in reversed(self.layers):
            z, ld = layer.inverse(z, t_cond, v_cond)
            log_det += ld
        return z, log_det

    def log_prob(self, x, t_cond, v_cond):
        z, log_det_inv = self.inverse(x, t_cond, v_cond)

        z_clipped = jnp.clip(z, -1.0, 1.0)
        log_pz = self.d * jnp.log(0.5)
        log_prob_final = log_pz + log_det_inv

        return log_prob_final

    def sample(self, key, n, t_cond, v_cond):
        z = jr.uniform(key, (n, self.d), minval=-1.0, maxval=1.0)
        x, _ = self.forward(z, t_cond, v_cond)
        return x
```

### 2.2. flow 물리 계산 및 결과

```python
# @title
class FreeEnergyEstimator:
    def __init__(self, flow_model: ConditionalRealNVP, gnn_model: LJPotential,Z, learning_rate: float):
        """
        Flow 모델, GNN 모델 객체를 저장하고 옵티마이저를 초기화한다.
        """
        self.flow_model = flow_model
        self.gnn_model = gnn_model
        self.Z=Z
        self.optimizer = nnx.Optimizer(flow_model, optax.adam(learning_rate), wrt=nnx.Param)

    def compute_zwanzig_free_energy(self, t_cond, v_cond, num_samples, key):
        """
        학습된 Flow 모델에서 num_samples만큼 x를 뽑은 뒤,
        GNN의 U(x)와 Flow의 log(q)섭동 식에 넣어 F 값 구하기
        """

        t_batch = jnp.full((num_samples,), t_cond)
        v_batch = jnp.full((num_samples,), v_cond)

        t_norm = ((t_batch - T_MIN)/(T_MAX-T_MIN)) * 2.0 - 1.0
        # 🚨 [수정] 여기서도 동일하게 로그 스케일 정규화
        log_v = jnp.log(v_batch)
        log_vmin = jnp.log(V_MIN)
        log_vmax = jnp.log(V_MAX)
        v_norm = ((log_v - log_vmin) / (log_vmax - log_vmin)) * 2.0 - 1.0

        x_samples = self.flow_model.sample(key, num_samples, t_norm,v_norm)

        L_box = v_cond ** (1/3)
        scale = L_box / 6.0

        log_q_x_raw = self.flow_model.log_prob(x_samples,t_norm,v_norm)

        log_q_x = log_q_x_raw - (x_samples.shape[-1] * jnp.log(scale))

        u_energy = jax.vmap(self.gnn_model,in_axes=(0, None,0,None))(x_samples,self.Z,v_batch,1.0)
        # ---------------------------------------------------------
        # 🚨 [추가: 꼬리 자르기 (Outlier Rejection)]
        # 에너지가 너무 높은 상위 1%의 샘플은 과감하게 버립니다.
        # ---------------------------------------------------------
        # u_energy의 99백분위수(percentile) 계산
        p99_val = jnp.percentile(u_energy, 99.0)

        # p99보다 작은 정상적인 샘플들만 남기기 위한 마스크 생성
        valid_mask = u_energy < p99_val

        # 유효한 샘플의 개수를 다시 셈
        valid_count = jnp.sum(valid_mask)

        # 마스크를 씌우지 않으면 형태가 깨지므로,
        # 쓸모없는 값은 계산에서 완전히 배제되도록 exponent를 극단적인 음수로 날려버림
        # (exp(-무한대) = 0 이므로 logsumexp에서 무시됨)
        u_scaled = u_energy / (KB * t_cond)
        exponent = jnp.where(valid_mask, -u_scaled - log_q_x, -1e10)

        # ---------------------------------------------------------
        # [2] 형상 자유 에너지 (Configurational Free Energy) 계산
        # ---------------------------------------------------------
        f_conf = -(KB * t_cond) * (jax.nn.logsumexp(exponent) - jnp.log(valid_count))

        # ---------------------------------------------------------
        # [3] 이상 기체 운동 에너지 및 구별 불가능성 (Kinetic & Gibbs Factor)
        # ---------------------------------------------------------
        N = self.Z.shape[0]
        t_val = jnp.squeeze(t_cond)
        f_kin = N * KB * t_val * (jnp.log(N) - 1.0 - 1.5 * jnp.log(t_val))

        # 총 헬름홀츠 자유 에너지는 이 둘의 합
        return f_conf + f_kin

    def train_step(self, z, t_cond, v_cond, epoch, n_epochs):
        # 현재 훈련 진행도를 0.0 ~ 1.0 사이의 실수로 계산
        progress = epoch / n_epochs

        # 내부 JIT 코어에 progress 전달
        loss, diag = _pure_jit_train_core(
            self.flow_model, self.optimizer, self.gnn_model,
            self.Z, z, t_cond, v_cond, progress
        )
        return loss, diag

@nnx.jit
def _pure_jit_train_core(flow_model, optimizer, gnn_model, Z, z, t_cond, v_cond, progress):
    t_norm = ((t_cond - T_MIN)/(T_MAX-T_MIN)) * 2.0 - 1.0
    # 🚨 [수정] V를 쌩으로 주지 않고, 로그 공간에서 정규화하여 전달
    log_v = jnp.log(v_cond)
    log_vmin = jnp.log(V_MIN)
    log_vmax = jnp.log(V_MAX)
    v_norm = ((log_v - log_vmin) / (log_vmax - log_vmin)) * 2.0 - 1.0

    def loss_fn(model):
        x, log_det = model.forward(z, t_norm, v_norm)

        v_cond_flat = jnp.reshape(v_cond, (-1,))
        L_box = v_cond_flat ** (1/3)
        scale = L_box / 6.0 #z=3(99.7%)일때 L/2(벽)에 위치함.

        x_physical = x * scale[:,None]
        D_DIM = x.shape[-1]
        log_det_physical = log_det + D_DIM * jnp.log(scale)

        u_energy = jax.vmap(gnn_model, in_axes=(0, None, 0, None))(x_physical, Z, v_cond, progress)
        u_scaled = u_energy / (KB * jnp.squeeze(t_cond))

        diff = x[:, None, :] - x[None, :, :]
        r2 = jnp.sum(diff**2, axis=-1)
        mask = 1.0 - jnp.eye(x.shape[0])

        r = jnp.sqrt(r2 + 1e-6)
        r_min = 0.8  # sigma 기준

        geom_penalty = jnp.mean(mask * jnp.maximum(r_min - r, 0.0)**2)
        geom_penalty=0#이상기체인 경우만.
        loss = (u_scaled - log_det_physical).mean()+0.01*geom_penalty

        #출력용
        loss_u = u_scaled.mean()

        diag = {
            "u_mean": u_energy.mean(),
            "u_max": u_energy.max(),
            "u_min": u_energy.min(),
            "logdet_mean": log_det_physical.mean(),
            "loss_u": loss_u ,
        }
        return loss, diag

    # has_aux=True를 켜서 보조 지표(diag)를 함께 반환받음
    (loss, diag), grads = nnx.value_and_grad(
        loss_fn, argnums=nnx.DiffState(0, nnx.Param), has_aux=True
    )(flow_model)

    optimizer.update(flow_model, grads)
    return loss, diag
```

```python

# @title
def run_free_energy_training(estimator, n_epochs, batch_size, T_bounds, V_bounds, key):
    t_min, t_max = T_bounds
    v_min, v_max = V_bounds
    losses = []

    for epoch in range(n_epochs):
        key, sub_z, sub_t, sub_v = jr.split(key, 4)

        z = jr.normal(sub_z, (batch_size, estimator.flow_model.d))
        t_cond = jr.uniform(sub_t, (batch_size, 1), minval=t_min, maxval=t_max)
        v_cond = jr.uniform(sub_v, (batch_size, 1), minval=v_min, maxval=v_max)

        # 🚨 epoch와 n_epochs 정보를 함께 넘겨줌
        loss, diag = estimator.train_step(z, t_cond, v_cond, epoch, n_epochs)
        losses.append(float(loss))

        if (epoch + 1) % 100 == 0:
            print(
                f"Epoch {epoch+1:4d} | Loss = {float(loss):.3f} | "
                f"U_mean = {float(diag['u_mean']):.1f} | U_max = {float(diag['u_max']):.1e} | "
                f"U_min = {float(diag['u_min']):.1f}"
                f"LogDet_mean = {float(diag['logdet_mean']):.3f}"
                f"Loss_u = {float(diag['loss_u']):.3f}"
            )

    return losses, key
```

## 3. F에서 메트릭($g_{\mu\nu}$) 찾고, geodesic 미방 풀기

```python
# @title
class GeodesicPathOptimizer(nnx.Module):
    def __init__(self, f_manifold_fn, num_points: int,
                 cycle_center: tuple, t_bounds: tuple, v_bounds: tuple, learning_rate: float):
        """
        :param f_manifold_fn: T, V를 입력받아 스칼라 F를 반환하는 함수
        :param num_points: 닫힌 루프에서 시작점을 제외한 가변 중간 점들의 개수
        :param cycle_center: 초기 원형 경로의 중심 좌표 (T_center, V_center)
        :param t_bounds: 전체 온도 영역 (T_min, T_max) -> 반지름 스케일링용
        :param v_bounds: 전체 부피 영역 (V_min, V_max) -> 반지름 스케일링용
        """
        self.f_manifold_fn = f_manifold_fn
        self.num_points = num_points
        self.center_node = jnp.array(cycle_center)

        self.t_min, self.t_max = t_bounds
        self.v_min, self.v_max = v_bounds
        self.grid_size = 150

        # 매니폴드 영역 크기의 10% 스케일로 반지름 설정
        r_t = (t_bounds[1] - t_bounds[0]) * 0.1
        r_v = (v_bounds[1] - v_bounds[0]) * 0.1

        #초기점(r_t,0) 설정 및 theta 세팅
        self.start_node = jnp.array([self.center_node[0] + r_t, self.center_node[1]])
        theta = jnp.linspace(0.0, 2.0 * jnp.pi, num_points + 1, endpoint=False)[1:]

        # 원형 경로로 중간 점들 초기화
        init_t = self.center_node[0] + r_t * jnp.cos(theta)
        init_v = self.center_node[1] + r_v * jnp.sin(theta)
        init_path = jnp.stack([init_t, init_v], axis=-1)

        self.mid_points = nnx.Param(init_path)

        self.f_grid = self._build_f_grid()
        self.g_grid = self._build_metric_grid()
        self.pressure_grid = self._build_pressure_grid()

        self.optimizer = nnx.Optimizer(self, optax.adam(learning_rate), wrt=nnx.Param)

    def _build_f_grid(self):
        t_coords = jnp.linspace(self.t_min, self.t_max, self.grid_size)
        v_coords = jnp.linspace(self.v_min, self.v_max, self.grid_size)

        rows = [
            jax.vmap(lambda v: self.f_manifold_fn(t, v))(v_coords)
            for t in t_coords
        ]

        return jnp.stack(rows, axis=0)

    def _build_metric_grid(self, eps=1e-8):
        dt = (self.t_max - self.t_min) / (self.grid_size - 1)
        dv = (self.v_max - self.v_min) / (self.grid_size - 1)

        dF_dT = jnp.gradient(self.f_grid, dt, axis=0)
        dF_dV = jnp.gradient(self.f_grid, dv, axis=1)

        gTT = jnp.gradient(dF_dT, dt, axis=0)
        gTV = jnp.gradient(dF_dT, dv, axis=1)
        gVT = jnp.gradient(dF_dV, dt, axis=0)
        gVV = jnp.gradient(dF_dV, dv, axis=1)

        metric = jnp.stack([
            jnp.stack([gTT, gTV], axis=-1),
            jnp.stack([gVT, gVV], axis=-1)
        ], axis=-2)

        return metric + eps * jnp.eye(2)

    def _build_pressure_grid(self):
        dv = (self.v_max - self.v_min) / (self.grid_size - 1)
        return -jnp.gradient(self.f_grid, dv, axis=1)

    def _get_metric_tensor(self, t, v, eps=1e-8):
        t_idx = (t - self.t_min) / (self.t_max - self.t_min) * (self.grid_size - 1)
        v_idx = (v - self.v_min) / (self.v_max - self.v_min) * (self.grid_size - 1)
        coords = jnp.stack([t_idx, v_idx])

        # g_grid[T, V, i, j] 에서 각 성분별로 보간 값을 뽑아냄
        g00 = jnd.map_coordinates(self.g_grid[:, :, 0, 0], coords, order=1, mode='nearest')
        g01 = jnd.map_coordinates(self.g_grid[:, :, 0, 1], coords, order=1, mode='nearest')
        g10 = jnd.map_coordinates(self.g_grid[:, :, 1, 0], coords, order=1, mode='nearest')
        g11 = jnd.map_coordinates(self.g_grid[:, :, 1, 1], coords, order=1, mode='nearest')

        g_off = 0.5 * (g01 + g10)

        metric = jnp.array([
            [g00, g_off],
            [g_off, g11]
        ])

        # 음수 없애기
        eigvals, eigvecs = jnp.linalg.eigh(metric)

        eigvals = jnp.maximum(eigvals,1e-4)

        metric = eigvecs @ jnp.diag(eigvals) @ eigvecs.T
        return metric + eps * jnp.eye(2)

    def _calculate_path_metrics(self, path):

        # segment(각 점 사이의 선분) 구하기
        dt_dv = jnp.diff(path, axis=0)
        seg_centers = path[:-1] + 0.5 * dt_dv

        g_matrices = jax.vmap(lambda pos: self._get_metric_tensor(pos[0], pos[1]))(seg_centers)

        # ds^2 = g_ij * dx^i * dx^j
        g_dx = jnp.einsum('nij,nj->ni', g_matrices, dt_dv)
        ds_squared = jnp.einsum('ni,ni->n', g_dx, dt_dv)

        # 총 Thermodynamic Length L
        L = jnp.sum(jnp.sqrt(jnp.clip(ds_squared, a_min=1e-7)))

        return ds_squared, L, dt_dv

    def _compute_path_length_loss(self, mid_points):
        total_path = jnp.concatenate([
            self.start_node[None, :],
            mid_points,
            self.start_node[None, :]
        ], axis=0)

        # 경로 길이 L loss
        ds_squared, L, dt_dv = self._calculate_path_metrics(total_path)

        # 점 뭉침 방지 loss
        euclidean_dists = jnp.sqrt(jnp.sum(dt_dv**2, axis=-1) + 1e-7)
        spacing_penalty = jnp.var(euclidean_dists) * 20.0

        return L + spacing_penalty

    def train_step(self):
        loss = _pure_jit_geodesic_core(self,self.optimizer)
        return loss

    def fit(self, n_epochs: int):
        losses = []
        for epoch in range(n_epochs):
            loss_val = self.train_step()
            losses.append(float(loss_val))

            if (epoch + 1) % 100 == 0:
                print(f"Geodesic Cycle Epoch {epoch+1:4d} | Loss (Length) = {losses[-1]:.4f}")
        return losses

    def get_interpolated_path(self, num_dense_points: int = 1000):
        # 닫힌 루프 형태로 노드들 병합
        final_nodes = jnp.concatenate([
            self.start_node[None, :],
            self.mid_points,
            self.start_node[None, :]
        ], axis=0)

        n = final_nodes.shape[0]
        t_nodes = jnp.linspace(0.0, 1.0, n)
        t_dense = jnp.linspace(0.0, 1.0, num_dense_points)

        t_est = jnp.interp(t_dense, t_nodes, final_nodes[:, 0])
        v_est = jnp.interp(t_dense, t_nodes, final_nodes[:, 1])

        return jnp.stack([t_est, v_est], axis=-1)


    def compute_pressure(self, t, v):
        t_idx = (t - self.t_min) / (self.t_max - self.t_min) * (self.grid_size - 1)
        v_idx = (v - self.v_min) / (self.v_max - self.v_min) * (self.grid_size - 1)

        coords = jnp.stack([t_idx, v_idx])

        return jnd.map_coordinates(
            self.pressure_grid,
            coords,
            order=1,
            mode='nearest'
        )

    def calculate_max_power(self, tau: float, tau_relax: float, num_dense_points: int = 1000):
        """ 경로를 따라 일(W_net)과 소산을 계산하여 최대 출력을 도출한다. """
        geodesic_path = self.get_interpolated_path(num_dense_points)
        T_path = geodesic_path[:, 0]
        V_path = geodesic_path[:, 1]

        P_path = jax.vmap(self.compute_pressure, in_axes=(0, 0))(T_path, V_path)

        # 닫힌 함수에 대해서는 사다리꼴 공식을 사용하면 오차 매우 작음
        W_net = -jnp.sum((P_path[:-1] + P_path[1:]) * 0.5 * jnp.diff(V_path))

        _, L, _ = self._calculate_path_metrics(geodesic_path)

        # [물리적 소산 에너지(Lost Work) 계산]
        # 유한 시간 열역학의 선형 응답 이론(Linear Response Theory)에 따름.
        # 1) 계수 0.5: 평형 근처 자유에너지(F)의 2차 테일러 전개(Quadratic Approximation)에서 기인.
        # 2) tau_relax: 순수 Hessian 메트릭(g)을 실제 마찰 텐서(Friction)로 연결하는 완화 시간 기여.
        # 3) 차원 정합성: W_net(Joule)과 단위를 맞추기 위해 무차원 시간 비율(tau_relax / tau)이 L^2에 곱해짐.
        dissipation = 0.5 * (tau_relax * (L ** 2)) / tau
        max_power = (W_net - dissipation) / tau

        return max_power, W_net, L, dissipation

    def plot_pv_diagram(self, num_dense_points: int = 1000):
        """ T-V 공간의 측지선 궤적을 P-V 좌표계로 변환하여 매끄러운 다이어그램을 그린다. """
        geodesic_path = self.get_interpolated_path(num_dense_points)
        T_path = geodesic_path[:, 0]
        V_path = geodesic_path[:, 1]

        P_path = jax.vmap(self.compute_pressure, in_axes=(0, 0))(T_path, V_path)

        plt.figure(figsize=(8, 6))
        plt.plot(V_path, P_path, marker='.', linestyle='-', color='teal', label='Geodesic Cycle', linewidth=1.5)
        plt.plot(V_path[0], P_path[0], marker='*', color='red', markersize=12, label='Start Point')

        plt.title("P-V Diagram of the Geodesic Cycle", fontsize=14)
        plt.xlabel("Volume ($V$)", fontsize=12)
        plt.ylabel("Pressure ($P$)", fontsize=12)
        plt.grid(True, linestyle='--', alpha=0.7)
        plt.legend()
        plt.tight_layout()
        plt.show()


@nnx.jit
def _pure_jit_geodesic_core(geo_model,optimizer):
    def loss_fn(model):
        return model._compute_path_length_loss(model.mid_points)


    loss, grads = nnx.value_and_grad(
        loss_fn,
        argnums=nnx.DiffState(0, nnx.Param)
    )(geo_model)

    optimizer.update(geo_model,grads)
    return loss
```

효율이 Curzon-Ahlborn efficiency 근처에서 노는지 확인

MD simulation에서 뽑은 relaxation time이 cycle 시간보다 크면 invalid

```python
# @title
## 5. 결과 검증

def compute_entropy(f_manifold_fn, t, v) -> float:
    """
    F를 온도 T로 편미분하여 해당 좌표의 엔트로피 S를 구한다. (S = -∂F/∂T)
    """
    grad_F_t = jax.grad(f_manifold_fn, argnums=0)(t, v)
    return -grad_F_t

def verify_cycle_performance(geodesic_path, f_manifold_fn, tau, T_hot_reservoir, T_cold_reservoir, tau_relax):
    """
    :param geodesic_path: 측지선 최적화로 구한 T, V 경로 데이터 (N, 2)
    :param f_manifold_fn: 자유에너지 매니폴드 함수 F(T, V)
    :param tau: 전체 사이클 시간 (조작 시간)
    :param T_hot_reservoir: 고온 열원의 실제 온도 (Th)
    :param T_cold_reservoir: 저온 열원의 실제 온도 (Tc)
    :param tau_relax: MD에서 추출한 기체 시스템의 완화 시간 (Relaxation time)
    """

    # 1. 4번 파트 함수를 활용하여 알짜 일과 소산 에너지 계산
    max_power, W_net, L, dissipation = calculate_max_power(geodesic_path, f_manifold_fn, tau)
    W_actual = W_net - dissipation

    # 2. 경로 상에서 흡수된 총 열량 (Q_in = ∮ T dS) 계산
    T_path = geodesic_path[:, 0]
    V_path = geodesic_path[:, 1]

    # 각 점에서의 엔트로피 계산 (vmap 활용)
    S_path = jax.vmap(compute_entropy, in_axes=(None, 0, 0))(f_manifold_fn, T_path, V_path)

    # 미소 엔트로피 변화량 dS 및 구간 중점의 온도 T_mid 계산
    dS = jnp.diff(S_path)
    T_mid = 0.5 * (T_path[:-1] + T_path[1:])
    dQ = T_mid * dS

    # 열기관이 외부 고온 열원으로부터 흡수한 총 열량 (dQ > 0 인 구간의 합산)
    Q_in = jnp.sum(jnp.clip(dQ, a_min=0.0))

    # 3. 효율 계산 (실제 효율 vs Curzon-Ahlborn 효율)
    efficiency = W_actual / (Q_in + 1e-7)
    eta_CA = 1.0 - jnp.sqrt(T_cold_reservoir / T_hot_reservoir)  # [수정] 외부 열원 온도로 정확히 계산
    eta_Carnot = 1.0 - (T_cold_reservoir / T_hot_reservoir)      # 가역 한계인 카르노 효율도 함께 계산

    # 4. 검증 리포트 출력
    print("=" * 20 + " [5. 결과 및 물리적 유효성 검증] " + "=" * 20)
    print(f"[1] 열역학적 효율 및 출력 분석")
    print(f"    - 외부 고온 열원 온도 (T_H)   : {T_hot_reservoir:.2f} K")
    print(f"    - 외부 저온 열원 온도 (T_C)   : {T_cold_reservoir:.2f} K")
    print(f"    - 작동 기체의 실제 온도 범위   : {jnp.min(T_path):.2f} K ~ {jnp.max(T_path):.2f} K")
    print(f"    - 최대 일률 상태의 효율 (η)    : {efficiency:.4f}")
    print(f"    - Curzon-Ahlborn 이론 효율 (η_CA): {eta_CA:.4f}")
    print(f"    - 이상적 카르노 이론 효율 (η_Carnot): {eta_Carnot:.4f}")
    print(f"    - 이론값과의 예측 편차 (|η - η_CA|) : {jnp.abs(efficiency - eta_CA):.4f}")
    print(f"      * 설명: 측지선 경로를 통한 일률 최적화 효율(η)이 CA 효율 근처에서 노는지 확인합니다.")

    print(f"\n[2] 준정적(Quasi-static) 가정 시계열 유효성 검증")
    print(f"    - 설정된 전체 사이클 조작 시간 (τ) : {tau:.4e} s")
    print(f"    - 분자 동역학(MD) 계 완화 시간 (τ_relax) : {tau_relax:.4e} s")

    if tau > tau_relax:
        print("    -> [RESULT: VALID] τ > τ_relax 상태입니다.")
        print("       매니폴드 표면의 미분기하학적 제어(Local Equilibrium) 가정이 물리학적으로 완전히 유효합니다.")
    else:
        print("    -> [RESULT: INVALID] τ <= τ_relax 상태입니다.")
        print("       조작 속도가 분자 완화 속도보다 빨라 정적 파라미터(자유에너지 표면) 정의가 무너집니다. τ를 키워야 합니다.")

    return efficiency, eta_CA
```

## 6. main

```python
# @title
def main():
    print("=" * 20 + " [6. 전체 파이프라인 컴파일 테스트] " + "=" * 20)

    # -------------------------------------------------------------
    # Step 0. 환경 변수 및 난수 키 설정
    # -------------------------------------------------------------
    master_key = jr.PRNGKey(42)
    master_key, flow_rng, gnn_rng, train_rng, path_rng = jr.split(master_key, 5)

    NUM_PARTICLES = 32
    INPUT_DIM = NUM_PARTICLES * 3  # 96 차원 세팅

    # 학습할 온도(T) 및 부피(V) 매니폴드 영역 선언
    T_BOUNDS = (0.5,5.0)
    V_BOUNDS = (1.0, 5.0)

    # -------------------------------------------------------------
    # Step 1. 모델 객체 초기화 (Flow & SchNet GNN)
    # -------------------------------------------------------------
    print("\n[Step 1] 머신러닝 모델 객체 초기화 중...")

    # 가본적인 조건부 RealNVP 생성 (d=96, 조건차원=2)
    flow_model = ConditionalRealNVP(
        d=INPUT_DIM, n_layers=12, hidden=64, cond_dim=2, rngs=nnx.Rngs(flow_rng)
    )

    # 🚨 [신규] 원자 번호 배열 Z 세팅 (예: 32개의 수소 원자로 가정)
    # 실제 데이터셋 적용 시 이 배열에 탄소(6), 산소(8) 등의 값을 넣으면 됩니다.
    Z_array = jnp.ones(NUM_PARTICLES, dtype=jnp.int32)

    # 🚨 [수정] SchNet 포텐셜 생성 (미분 가능한 SiLU 적용)
    gnn_model = LJPotential(
    sigma=1.0,
    epsilon=1.0,
    wall_k=20.0
    )

    # 학습 매니저 객체 바인딩 (Z 배관 추가)
    estimator = FreeEnergyEstimator(
        flow_model=flow_model,
        gnn_model=gnn_model,
        Z=Z_array,
        learning_rate=1e-3
    )
    print(" -> 모델 초기화 완료.")

    # -------------------------------------------------------------
    # Step 2. 생성 모델(Flow) 기반 자유 에너지 매니폴드 F(T, V) 학습
    # -------------------------------------------------------------
    print("\n[Step 2] 볼츠만 분포 역방향 KL 발산 학습 시작...")
    # 32개 입자는 가볍기 때문에 컴파일 및 학습 속도가 매우 빠릅니다.
    n_epochs = 2000
    batch_size = 256

    losses, next_key = run_free_energy_training(
        estimator=estimator,
        n_epochs=n_epochs,
        batch_size=batch_size,
        T_bounds=T_BOUNDS,
        V_bounds=V_BOUNDS,
        key=train_rng
    )
    print(" -> 매니폴드 맵핑 학습 완료.")

    # -------------------------------------------------------------
    # Step 3. 섭동 식 기반의 연속적 F(T, V) 대리 함수 선언
    # -------------------------------------------------------------
    print("\n[Step 3] Zwanzig 섭동 식 기반의 미분 가능한 F(T, V) 지형 함수 빌딩 중...")

    # 수치 미분(jax.hessian)에 완벽하게 대응하기 위해
    # 고정된 난수 시드와 대량 샘플(1000개)을 활용해 래핑된 순수 함수 선언
    def final_f_manifold(t, v):
        # 학습에 쓰이지 않은 고정된 내부 평가용 키 주입
        eval_key = jr.PRNGKey(127)
        return estimator.compute_zwanzig_free_energy(
            t_cond=t,
            v_cond=v,
            num_samples=128,
            key=eval_key
        )

    # 지형 함수가 깨지지 않고 미분 값을 잘 뱉는지 정상 유무 사전 체크
    test_t, test_v = 3.0, 3.0
    test_f = final_f_manifold(test_t, test_v)
    print(f" -> 검증 점 평가 완료: F({test_t} K, {test_v} L) = {test_f:.4e} Joule")

    # -------------------------------------------------------------
    # Step 4. 매니폴드 위에서 ds^2 최소화를 통한 Geodesic 폐곡선 탐색
    # -------------------------------------------------------------
    print("\n[Step 4] $ds^2$ 최소화 기반의 열역학적 최적 측지선(Geodesic) 고무줄 당기기 시작...")

    # 매니폴드 평면의 중심을 기준으로 도넛 모양 고무줄 초기 배치 설정
    CYCLE_CENTER = (1.5,50)

    path_optimizer = GeodesicPathOptimizer(
        f_manifold_fn=final_f_manifold,
        num_points=40,               # 40개의 가변 마디 점 생성 (시작점 제외)
        cycle_center=CYCLE_CENTER,
        t_bounds=T_BOUNDS,
        v_bounds=V_BOUNDS,
        learning_rate=5e-3
    )

    # 고무줄을 매니폴드의 곡률 곡면에 맞춰 팽팽하게 수렴시킴
    path_losses = path_optimizer.fit(n_epochs=400)
    print(" -> 최적화 측지선 루프 획득 완료.")

    # -------------------------------------------------------------
    # Step 5. 물리 지표 연산 및 PV 다이어그램 시각화 투영
    # -------------------------------------------------------------
    print("\n[Step 5] 유한 시간 열역학 기반 출력(Max Power) 연산 및 PV 다이어그램 시각화...")

    # 가상의 조작 조건 상수 주입 (전체 조작 시간 tau, 계의 완화 시간 tau_relax)
    TAU = 1.0e-2
    TAU_RELAX = 1.0e-4

    max_power, w_net, length, dissipation = path_optimizer.calculate_max_power(
        tau=TAU,
        tau_relax=TAU_RELAX,
        num_dense_points=500  # 500개의 촘촘한 보간점들로 고해상도 연산
    )

    print("-" * 50)
    print(f" * 알짜 일 (W_net)          : {w_net:.4e} J")
    print(f" * 기하학적 길이 (Length L)  : {length:.4e}")
    print(f" * 열적 소산 (Dissipation)   : {dissipation:.4e} J")
    print(f" * 최종 최대 출력 (Max Power): {max_power:.4e} W")
    print("-" * 50)

    # P-V 평면 상에 최종 곡선 렌더링 출력
    path_optimizer.plot_pv_diagram(num_dense_points=500)
    print("\n=== 전체 파이프라인 정상 작동 확인 완료! ===")

if __name__ == "__main__":
    main()
```

```python
# @title
import jax
import jax.numpy as jnp
import jax.random as jr
import flax.nnx as nnx

def run_unit_tests():
    print("=" * 40)
    print(" 🛠️ 개별 모듈 유닛 테스트 (Unit Tests) 🛠️")
    print("=" * 40)

    # 공통 환경 세팅
    key = jr.PRNGKey(42)
    k_flow, k_gnn, k_data = jr.split(key, 3)

    NUM_PARTICLES = 32
    INPUT_DIM = NUM_PARTICLES * 3  # 96
    BATCH_SIZE = 128

    # -----------------------------------------------------------------
    # [Test 1] GNN 포텐셜 모듈 단독 테스트
    # -----------------------------------------------------------------
    print("\n[Test 1] GNNPotential 단독 검증 시작...")
    try:
        gnn = GNNPotential(hidden_dim=32, num_layers=2, rngs=nnx.Rngs(k_gnn))

        # 가상의 분자 배치 생성 (BATCH_SIZE, 96)
        dummy_x = jr.normal(k_data, (BATCH_SIZE, INPUT_DIM))

        # 1-1. 순전파 테스트 (에너지 출력)
        u_out = jax.vmap(gnn)(dummy_x)
        print(f"  ✅ 순전파 성공: 에너지 출력 형태 {u_out.shape}, 평균값: {u_out.mean():.4f}")

        # 1-2. 미분(Gradient) 테스트 (NaN 발생 유무 체크)
        # 에너지를 x로 미분했을 때 힘(Force) 벡터가 무사히 나오는지 확인
        grad_fn = jax.vmap(jax.grad(gnn))
        forces = grad_fn(dummy_x)

        if jnp.isnan(forces).any():
            print("  ❌ [실패] GNN 미분 시 NaN이 발생합니다! (거리 계산 로직 확인 필요)")
        else:
            print(f"  ✅ 미분 성공: NaN 없음. 힘 벡터 형태 {forces.shape}")

    except Exception as e:
        print(f"  ❌ GNN 테스트 중 에러 발생: {e}")

    # -----------------------------------------------------------------
    # [Test 2] Conditional Flow 모듈 단독 테스트
    # -----------------------------------------------------------------
    print("\n[Test 2] ConditionalRealNVP 단독 검증 시작...")
    try:
        flow = ConditionalRealNVP(d=INPUT_DIM, n_layers=8, hidden=64, cond_dim=2, rngs=nnx.Rngs(k_flow))

        # 가상의 입력 데이터
        dummy_z = jr.normal(k_data, (BATCH_SIZE, INPUT_DIM))
        dummy_t = jnp.ones((BATCH_SIZE, 1)) * 3.0
        dummy_v = jnp.ones((BATCH_SIZE, 1)) * 3.0

        # 2-1. 조건부 순전파 테스트
        x_out, log_det = flow.forward(dummy_z, dummy_t, dummy_v)
        if jnp.isnan(x_out).any() or jnp.isnan(log_det).any():
            print("  ❌ [실패] Flow 순전파 중 NaN 발생!")
        else:
            print(f"  ✅ 순전파 성공: x 형태 {x_out.shape}, log_det 형태 {log_det.shape}")

        # 2-2. 역전파(Inverse) 완벽 복원 테스트
        # x_out을 다시 inverse에 넣었을 때 원래의 dummy_z와 완벽히 똑같아야 함
        z_back, log_det_inv = flow.inverse(x_out, dummy_t, dummy_v)

        # 복원 오차율 계산 (기계 정밀도 내에서 0에 가까워야 함)
        max_error = jnp.max(jnp.abs(dummy_z - z_back))

        if max_error > 1e-4:
            print(f"  ❌ [실패] Flow 가역성 깨짐! (최대 복원 오차: {max_error:.2e})")
        else:
            print(f"  ✅ 가역성(Inverse) 성공: 최대 복원 오차 {max_error:.2e} (수학적으로 완벽함)")

        # 2-3. Log Prob 확률 연산 테스트
        log_probs = flow.log_prob(x_out, dummy_t, dummy_v)
        if jnp.isnan(log_probs).any():
            print("  ❌ [실패] log_prob 연산 중 NaN 발생!")
        else:
            print(f"  ✅ 확률 밀도(log_prob) 연산 성공: 결과 형태 {log_probs.shape}")

    except Exception as e:
        print(f"  ❌ Flow 테스트 중 에러 발생: {e}")

    print("\n=== 모든 개별 유닛 테스트 종료 ===")

run_unit_tests()
```

```python
# @title
import jax
import jax.numpy as jnp
import numpy as np

# 🚨 질문자님의 환경에 맞게 키(key)와 바운드(v_norm 계산)를 맞춰주세요!
# v_norm_offset = V_MIN, v_norm_scale = (V_MAX - V_MIN)
def check_out_of_box_fraction(estimator, t_vals, v_vals, num_samples, key):
    print("==================================================")
    print(" 🕵️ 진단 테스트: 박스 이탈률 (Out-of-Box Fraction) 측정")
    print("==================================================")

    N_PARTICLES = 32
    oob_grid = np.zeros((len(t_vals), len(v_vals)))

    for i, t in enumerate(t_vals):
        for j, v in enumerate(v_vals):
            # 1. 조건 세팅
            t_batch = jnp.full((num_samples, 1), float(t))
            v_batch = jnp.full((num_samples, 1), float(v))

            t_norm = ((t_batch - T_MIN) / T_MAX) * 2.0 - 1.0
            # 🚨 주의: 아래 정규화 식은 실제 V_BOUNDS에 맞게 꼭 수정하세요! (예: 30~300)
            v_norm = ((v_batch - V_MIN) / V_MAX) * 2.0 - 1.0

            # 2. Flow 샘플링
            key, subkey = jax.random.split(key)
            x_flow = estimator.flow_model.sample(subkey, num_samples, t_norm, v_norm)

            # 수치 안정화 (이전 패치 유지)
            x_flow = jnp.nan_to_num(x_flow, nan=0.0, posinf=10.0, neginf=-10.0)

            # 3. 물리적 스케일링
            v_batch_flat = jnp.reshape(v_batch, (-1,))
            L_box = v_batch_flat ** (1/3)
            scale = L_box / 6.0
            x_physical = x_flow * scale[:, None]

            # 4. 박스 이탈 판정 (Wall Boundary: ± L_box / 2)
            # x_physical: (batch, 96) -> (batch, 32, 3)으로 변환
            x_3d = jnp.reshape(x_physical, (num_samples, N_PARTICLES, 3))

            # half_L: (batch, 1, 1) 형태로 브로드캐스팅 맞춤
            half_L_3d = (L_box / 2.0)[:, None, None]

            # 단 하나의 입자, 단 하나의 축이라도 벽을 넘었는가?
            is_out = jnp.any(jnp.abs(x_3d) > half_L_3d, axis=(1, 2))

            # 이탈한 샘플의 비율(%) 계산
            oob_fraction = jnp.mean(is_out)
            oob_grid[i, j] = float(oob_fraction)

    # --- 결과 텍스트 표(Table) 출력 ---
    header = f"{'T \\ V':<6} |"
    for v in v_vals:
        header += f"{v:>6.1f}"
    print(header)
    print("-" * len(header))

    for i, t in enumerate(t_vals):
        row_str = f"{t:6.2f} |"
        for j in range(len(v_vals)):
            # 보기 쉽게 퍼센트(%)로 출력
            row_str += f"{oob_grid[i, j]*100:>5.1f}% "
        print(row_str)

    return oob_grid

# 실행 예시 (평가용으로 t_vals, v_vals 몇 개만 추려서)
# test_t_vals = np.linspace(0.5, 3.0, 5)
# test_v_vals = np.linspace(30.0, 300.0, 5)
# check_out_of_box_fraction(estimator, test_t_vals, test_v_vals, 1000, train_rng)
```

```python
import jax.random as jr
import jax.numpy as jnp
import numpy as np
import matplotlib.pyplot as plt

print("=" * 50)
print(" 🛠️ [1단계] 모델 초기화 및 훈련소 입소")
print("=" * 50)

# 기본 세팅
master_key = jr.PRNGKey(42)
master_key, flow_rng, gnn_rng, train_rng = jr.split(master_key, 4)
NUM_PARTICLES = 32
T_MIN = 0.5; T_MAX = 2.5
V_MIN = 30.0; V_MAX = 100.0

# Flow 및 GNN 모델 실물 생성
flow_model = ConditionalRealNVP(
    d=NUM_PARTICLES * 3, n_layers=16, hidden=128, cond_dim=2, rngs=nnx.Rngs(flow_rng)
)
Z_array = jnp.ones(NUM_PARTICLES, dtype=jnp.int32)

# 🚨 이상기체 테스트를 위해 입자간 상호작용(epsilon)을 0으로 끄고 벽(wall_k)만 남깁니다.
gnn_model = LJPotential(sigma=1.0, epsilon=0.0, wall_k=20.0)

estimator = FreeEnergyEstimator(
    flow_model=flow_model, gnn_model=gnn_model, Z=Z_array, learning_rate=1e-3
)

# 학습 시작 (이 과정을 거쳐야 모델이 물리 법칙을 깨닫습니다)
print("🚀 이상기체 상태(epsilon=0)로 학습을 시작합니다. (약 1~2분 소요 예상)")
losses, _ = run_free_energy_training(
    estimator=estimator,
    n_epochs=2000,          # 빠른 테스트를 위해 2000번만 돌립니다.
    batch_size=256,
    T_bounds=(T_MIN, T_MAX),
    V_bounds=(V_MIN, V_MAX),
    key=train_rng
)
print("✅ 물리 학습 완료!\n")

print("=" * 50)
print(" 🕵️ [2단계] 등온(T=1.0) 이상기체 F(V) 거동 분석 테스트")
print("=" * 50)

def final_f_manifold(t, v):
    eval_key = jr.PRNGKey(127)
    return estimator.compute_zwanzig_free_energy(
        t_cond=t, v_cond=v, num_samples=128, key=eval_key
    )

T_test = 1.0
v_test_vals = np.linspace(V_MIN, V_MAX, 30)
F_predicted = []
target_Vs = [30, 50, 80, 100]

print(f"--- T={T_test} 에서의 F(V) 수치 확인 ---")
for v in v_test_vals:
    f_val = float(final_f_manifold(T_test, v))
    F_predicted.append(f_val)
    if any(abs(v - target) < 1.0 for target in target_Vs):
        print(f"V = {v:5.1f} | 예측된 F(V) = {f_val:.4f}")

F_predicted = np.array(F_predicted)

# 이론적 이상기체 로그 곡선 (-NT ln V) 계산 및 평행이동(Shift)
KB = 1.0
F_ideal_raw = -NUM_PARTICLES * KB * T_test * np.log(v_test_vals)
shift_C = F_predicted[0] - F_ideal_raw[0]
F_ideal = F_ideal_raw + shift_C

# 결과 시각화
plt.figure(figsize=(8, 6))
plt.plot(v_test_vals, F_predicted, 'bo-', linewidth=2, label='Flow Model $F(V)$')
plt.plot(v_test_vals, F_ideal, 'r--', linewidth=2, label=r'Theoretical Ideal Gas ($-NT \ln V$)')

plt.title(f"Isothermal Free Energy $F(V)$ at $T={T_test}$", fontsize=14)
plt.xlabel("Volume ($V$)", fontsize=12)
plt.ylabel("Free Energy ($F$)", fontsize=12)
plt.legend(fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()

correlation = np.corrcoef(F_predicted, F_ideal)[0, 1]
print(f"-> 🎯 이론적 로그 곡선과의 형태 유사도(Correlation): {correlation:.4f} (1.0에 가까울수록 정상)")
```
